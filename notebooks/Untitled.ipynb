```python
fig, axes = plt.subplots(4,2, figsize=(12,12))
fig.suptitle('Distribuição das dos dados numéricos baseado nas classes')
fig.tight_layout(pad=3)
fig.supylabel('Total')
axes[0,0].set(ylabel="Total")
axes[0,1].set(ylabel="Total")
axes[1,0].set(ylabel="Total")
axes[1,1].set(ylabel="Total")
axes[2,0].set(ylabel="Total")
axes[2,1].set(ylabel="Total")
axes[3,0].set(ylabel="Total")
axes[3,1].set(ylabel="Total")
seaborn.histplot(ax=axes[0,0], x="G", data=data,hue="Glaucoma",bins=50)
seaborn.histplot(ax=axes[0,1], x="Idade", data=data,hue="Glaucoma",bins=50)
seaborn.histplot(ax=axes[1,0], x="T", data=data,hue="Glaucoma",bins=50)
seaborn.histplot(ax=axes[1,1], x="TS", data=data,hue="Glaucoma",bins=50)
seaborn.histplot(ax=axes[2,0], x="N", data=data,hue="Glaucoma",bins=50)
seaborn.histplot(ax=axes[2,1], x="NS", data=data,hue="Glaucoma",bins=50)
seaborn.histplot(ax=axes[3,0], x="NI", data=data,hue="Glaucoma",bins=50)
seaborn.histplot(ax=axes[3,1], x="TI", data=data,hue="Glaucoma",bins=50)
```

```python
fig, axes = plt.subplots(4,2, figsize=(12,12))
fig.suptitle('Boxplot Glaucoma ')
fig.tight_layout(pad=3)
seaborn.boxplot(ax=axes[0,0], x="G", data=data,y="Glaucoma")
seaborn.boxplot(ax=axes[0,1], x="Idade", data=data,y="Glaucoma")
seaborn.boxplot(ax=axes[1,0], x="T", data=data,y="Glaucoma")
seaborn.boxplot(ax=axes[1,1], x="TS", data=data,y="Glaucoma")
seaborn.boxplot(ax=axes[2,0], x="N", data=data,y="Glaucoma")
seaborn.boxplot(ax=axes[2,1], x="NS", data=data,y="Glaucoma")
seaborn.boxplot(ax=axes[3,0], x="NI", data=data,y="Glaucoma")
seaborn.boxplot(ax=axes[3,1], x="TI", data=data,y="Glaucoma")
```

```python
preprocessed_data = pd.read_csv("../data.csv")
```

```python
preprocessed_data
```

```python
preprocessed_data.columns
```

```python
preprocessed_data["Glaucoma"] = preprocessed_data["label"].map({1: "Sim", 0: "Não"})
preprocessed_data["Gênero"] = preprocessed_data["gender"].map({1: "Homem", 0: "Mulher"})
preprocessed_data["Idade"] = preprocessed_data["age"]
preprocessed_data["Classe"] = preprocessed_data["label"]
```

```python
data_cleaned = preprocessed_data[["Patient","Classe",'T/TS', 'T/NS', 'T/N', 'T/NI',
       'T/TI', 'TS/NS', 'TS/N', 'TS/NI', 'TS/TI', 'NS/N', 'NS/NI', 'NS/TI',
       'N/NI', 'N/TI', 'NI/TI']]
corr_df =  abs(data_cleaned.corr(method='pearson'))
```

```python
import numpy as np
df_lt = corr_df.where(np.tril(np.ones(corr_df.shape)).astype(np.bool))
```

```python
fig, ax = plt.subplots(figsize=(12,12))         # Sample figsize in inches

seaborn.heatmap(df_lt,annot=True,ax=ax,fmt='.1g')
```
