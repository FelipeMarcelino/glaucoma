```python
model.load_state_dict(torch.load("../models/default.pth"))
```

```python
dataloader_val = torch.load("../models/val_default_dataloader.pth")
```

```python
len(dataloader_val.dataset)
```

```python
num_train_examples = 105
background = Subset(dataloader_val.dataset, np.arange(num_train_examples))
```

```python
features_back = []
for i in range(len(background)):
    features_back.append(background[i][0])
```

```python
sample_train_shap = features_back[:100]
test_sample = features_back[100:]
```

```python
sample_train_shap[0].shape
```

```python
np.stack(sample_train_shap)
```

```python
e = shap.DeepExplainer(model, torch.from_numpy(np.stack(sample_train_shap)))
shap_values = e.shap_values(torch.from_numpy(np.stack(test_sample)))
```

```python
shap_numpy = [np.swapaxes(np.swapaxes(s, 1, -1), 1, 2) for s in shap_values]
test_numpy = np.swapaxes(np.swapaxes(np.stack(test_sample), 1, -1), 1, 2)
```

```python
shap.image_plot(shap_numpy, -test_numpy, width=30, aspect=0.1, hspace=0.1)
```
